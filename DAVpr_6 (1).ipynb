**Practical-6| Aim:Create Basic data visualization(Line,Bar,Histogram,pie,scatter)**

**step-1:import libraries**

```python
# Import Libraries.
# Import Pandas Library.
# Import NumPy Library.
# Import Matplot Library.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**step-2:create dataset**

```python
#create dataset of products
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product":["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Region":["North", "South", "East", "West", "North", "South", "East", "West", "East", "West", "West", "West"],
    "Sales":[120000, 135000, 150000, 145000, 160000, 175000, 180000,170000, 190000, 210000, 230000, 250000],
    "Units_sold":[12,30,25,14,35,30,18,38,32,21,45,40]
}
df = pd.DataFrame(data)
df
```

**step-3:create a linechart,
how did company sales change from jan to dec**

```python
#create a line chart
plt.figure(figsize=(10, 5))
plt.plot(df['Month'], df['Sales'], marker='o', color='b', linestyle='-', linewidth=2)
#give a title to line chart
plt.title('Monthly Sales Trend (Jan - Dec)')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid(True, linestyle='--', alpha=0.6)
#display the line chart
plt.show()
```

**step-4:create a barchart,which product generate how much sales**

```python
#make group of product and sales
product_sales = df.groupby('Product')['Sales'].sum().reset_index()

plt.figure(figsize=(8, 5))
plt.bar(product_sales['Product'], product_sales['Sales'], color=['skyblue', 'salmon', 'lightgreen'])
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**step-5:create a histogram, how are the sale distributed**

```python
#create a histogram
plt.figure(figsize=(8, 5))
plt.hist(df['Sales'], bins=5, color='lightblue', edgecolor='black')
plt.title('Distribution of Sales')
plt.xlabel('Sales ($)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--')
plt.show()
```

**step-6:create a scatter plot,is there a relationship between the number of unit sold and total sales**

```python
#create a scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df['Units_sold'], df['Sales'], color='darkorange', edgecolor='k')
plt.title('Relationship Between Units Sold and Total Sales')
plt.xlabel('Units Sold')
plt.ylabel('Sales ($)')
plt.grid(True, linestyle='--')
plt.show()
```

**step-7:create a piechart,what percentage of total sales comes from sales**

```python
#make a group of product and sales
product_sales = df.groupby('Product')['Sales'].sum()

plt.figure(figsize=(6, 6))
plt.pie(product_sales, labels=product_sales.index, autopct='%1.1f%%', startangle=140, colors=['skyblue', 'red', 'lightgreen'])
plt.title('Percentage of Total Sales by Product')
plt.show()
```

**Step-8: save dataset in csv file**

```python
#save data set in csv file
df.to_csv('sales_dataset.csv', index=False)
print("Dataset successfully saved as 'sales_dataset.csv'!")
```
