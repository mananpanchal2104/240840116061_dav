 **Practical-1 |Study the Data Analytics Lifecycle and perform data acquisition from csv,excel and json files using pandas**

```python
import pandas as pd
#data we create
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "krisha", "Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["Surat", "Navsari", "Vaslad", "Bharuch", "Mumbai", "Banglore", "Pune"],
    "Marks": [84, 78, 97, 69, 78, 80, 90]
}
```

```python
#convert dictionary into pandas dataframe
df = pd.DataFrame(data)
print(df)
```

```python
#save the dataframe as csv file
df.to_csv("students.csv", index=False)
#display the message
print("csv file is created")
```

```python
#save the dataframe as excel file
df.to_excel("students.xlsx", index=False)
#display the message
print("excel file is created")
```

```python
#save the dataframe as json file
df.to_json("students.json", index=False)
#display the message
print("json file is created")
```

```python
#display csv file
df_csv = pd.read_csv("students.csv")
print(df_csv)
```

```python
#display excel file
df_excel = pd.read_excel("students.xlsx")
print(df_excel)
```

```python
#display json file
df_json = pd.read_json("students.json")
print(df_json)
```
