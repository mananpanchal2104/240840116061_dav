step-1:import required labraries

```python
#import the pandas library
#pandas is used for reading , manipulation and analying datasets
import pandas as pd
#import the matplotlib library
import matplotlib.pyplot as plt
```

step-2:create dataset

```python
#creating dataset
Data = {
    "Month": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Units_sold": [120, 135, 150, 145, 160, 175, 180, 170, 190, 210, 230, 250],
    "Revenue": [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Profit": [20000, 23000, 25000, 24000, 27000, 30000, 31000, 29000, 33000, 37000, 41000, 45000]
}
df=pd.DataFrame(Data)
df
```

step-4:Generate Descriptive Statistic

```python
#Generate descriptive
df.describe()
```

step-5:calcualte mean

```python
#calcualte the avg numbers of unit sold per month.
average_units = df["Units_sold"].mean()
print("Average units sold per month:", average_units)
```

step-6:calcualte median

```python
#calculate the middle value pf the units_sold in month
median_units = df["Units_sold"].median()
print("Median units sold per month:", median_units)
```

step-7:calculate the max and min

```python
#Find the highest number of units sold in the month
max_units = df["Units_sold"].max()
print("Maximum units sold in a month:", max_units)

#Find the lowest number of units sold in month
min_units = df["Units_sold"].min()
print("Minimum units sold in a month:", min_units)
```

step-8:calculate standard deviation

```python
#calculate the standard deviation of units sold.
std_units = df["Units_sold"].std()
#Display the result.
print("Standard deviation of units sold:", std_units)
```

step-9:find the Month with highest sales

```python
#find the index of the row having the max revenue.
highest_sales_index = df["Revenue"].idxmax()

#Retrieve the month corresponding to that index.
highest_sales_month = df.loc[highest_sales_index, "Month"]

#Retrieve the maximum revenue
highest_sales_revenue = df.loc[highest_sales_index, "Revenue"]

#Display the result.
print("Month with highest sales:", highest_sales_month)
print("Highest sales revenue:", highest_sales_revenue)
```

step-10:Find the month with Lowest sales

```python
#find the index of the row having the min revenue.
lowest_sales_index = df["Revenue"].idxmin()

#Retrieve the month corresponding to that index.
lowest_sales_month = df.loc[lowest_sales_index, "Month"]

#Retrieve the minimum revenue
lowest_sales_revenue = df.loc[lowest_sales_index, "Revenue"]

#Display the result.
print("Month with lowest sales:", lowest_sales_month)
print("Lowest sales revenue:", lowest_sales_revenue)
```

step-11:Calculate Total Revenue

```python
#calcualte the total revenue generated during the year.
total_revenue = df["Revenue"].sum()
print("Total revenue generated:", total_revenue)
```

step-12:Calculate the total profit

```python
#calculate the total profit of the year
total_profit = df["Profit"].sum()
print("Total profit:", total_profit)
```

step-13:Identify the sales trend

```python
#create a line chart to visualize monthly revenue.
plt.figure(figsize=(10, 5))

#plot month on X-axis and revenue on the Y-axis.
plt.plot(df["Month"], df["Revenue"], marker="o")
#Add a title to graph.
plt.title("Monthly Revenue Trend")
#rotation of month label
plt.xticks(rotation=30)
#label x-axis month
plt.xlabel("Month")
#label y-axis revenue
plt.ylabel("Revenue")
#display the graph
plt.show()
```

step-14:Analyze the profit

```python
#create a linechart for monthly profit
plt.figure(figsize=(10, 5))

#plot month on X-axis and profit on the Y-axis.
plt.plot(df["Month"], df["Profit"], marker="o")
#Add a title to graph.
plt.title("Monthly Profit Trend")
#rotation of month label
plt.xticks(rotation=30)
#label x-axis month
plt.xlabel("Month")
#label y-axis profit
plt.ylabel("Profit")
#display the graph
plt.show()
```

step-15:compare revenue and profit

```python
#create a line chart to compare revenue and profit
plt.figure(figsize=(10, 5))

#plot month on X-axis and revenue on the Y-axis.
plt.plot(df["Month"], df["Revenue"], marker="o", label="Revenue")
#plot month on X-axis and profit on the Y-axis.
plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")
#Add a title to graph.
plt.title("Revenue vs Profit Comparison")
#rotation of month label
plt.xticks(rotation=30)
#display the legend
plt.legend()
#label x-axis month
plt.xlabel("Month")
#label y-axis revenue and profit
plt.ylabel("Amount")
#display the graph
plt.show()
```

step-16:identify growth i sales

```python
#calcualte the percentage change in revenue
#from one month to the next.

df["Revenue_Growth_%"] = df["Revenue"].pct_change() * 100
#display the result
df
```

step-17:identify months with increasing revenue

```python
#filter the dataset to display only those months
#where revenue growth is positive.
increasing_revenue_months = df[df["Revenue_Growth_%"] > 0]

#Display the result.
print("Months with increasing revenue:")
print(increasing_revenue_months)
```
