```python
#importing pandas library
import pandas as pd

data = {
    "ID" : [ 101, 102, 103, 104, 104, 105, 106, 107],
    "Name" : ["Rahul", "Priya", "Amit", None, "Riya", "Ajay", "Naresh", "Jenil"],
    "Age" : [20, 21, None, 20, 20, 22, 23, 28],
    "City" : ["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahmedabad"],
    "Marks" : [85, None, 78, 88, 88, 81, 90, 93]
}
df = pd.DataFrame(data)
print(df)
```

```python
#display the first 5 records of dataset
df.head()
```

```python
#display the last 5 records of dataset
df.tail()
```

```python
#display the detail infromation of dataset
df.info()
```

```python
#display the stastistical of dataset
df.describe()
```

```python
#display the dimension of dataset
df.shape
```

```python
#display the datatype of each column
df.dtypes
```

```python
#display all the column of dataset
df.columns
```

```python
#check missing values
df.isnull()
```

```python
#check missing values with number
df.isnull().sum()
```

```python
#print the total no. of null
df.isnull().sum().sum()
```

```python
#fill the missing values in marks column
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

```python
#replace the missing value in age column
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
#replace the null values in name column
df["Name"] = df["Name"].fillna("Manan")
df
```

```python
#detect duplicate
df.duplicated()
```

```python
#count the no. of duplicated rows
df.duplicated().sum()
```

```python
#remove the duplicated records
df.drop_duplicates()
```

```python
#rename the column "Marks" to "score".
df.rename(columns={"Marks":"Score"}, inplace=True)
df
```

```python
#convert the age column from float to integer
df["Age"] = df["Age"].astype(int)
display(df)
```

```python
#create a new column named Final score
df["Final_Score"] = df["Score"] + 5
display(df)
```

```python
#convert all student names into uppercase
df["Name"] = df["Name"].str.upper()
display(df)
```

```python
#convert all city names into lowercase
df["City"] = df["City"].str.lower()
display(df)
```

```python
#remove extra space from city column
df["City"] = df["City"].str.strip()
display(df)
```

```python
#remove extra space from name column from right
df["Name"] = df["Name"].str.rstrip()
display(df)
```

```python
#remove extra space from name column from left
df["Name"] = df["Name"].str.lstrip()
display(df)
```

```python
#sort the dataset based on final_score
df.sort_values(by="Final_Score", ascending=False)
```

```python
#save the cleaned dataset in csv
df.to_csv("cleaned_data.csv", index=False)
print("Data saved successfully")
```
